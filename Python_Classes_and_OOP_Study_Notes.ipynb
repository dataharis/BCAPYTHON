## 1. What is Object-Oriented Programming?

**Object-Oriented Programming (OOP)** organizes programs around objects that combine data and behavior.

- **Class:** a blueprint or template.
- **Object:** an instance created from a class.
- **Attributes:** data stored on a class or object.
- **Methods:** functions defined inside a class.

For example, `Student` can be a class, while individual students such as Aman and Neha are objects created from it.

### Advantages of OOP
- Reusability through inheritance and shared methods.
- Modularity: code is organized into logical units.
- Maintainability: easier to update and debug.
- Encapsulation: helps control access to data.
- Flexibility: different objects can provide different implementations of a common method.

## 2. Defining a class and creating objects

````python
class Student:
    college = "City College"  # Class attribute

s1 = Student()                # Create an object
s2 = Student()

print(s1.college)
print(s2.college)
````

**Expected output**
```text
City College
City College
```

`class Student:` defines a class. `Student()` creates an instance. `college` is a class attribute shared as a default by the instances.

### Example: Different objects with their own data

````python
class Mobile:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

m1 = Mobile("Samsung", "Galaxy")
m2 = Mobile("Apple", "iPhone")

print(m1.brand, m1.model)
print(m2.brand, m2.model)
````

**Expected output**
```text
Samsung Galaxy
Apple iPhone
```
Each object stores its own `brand` and `model` values.

## 3. The `__init__()` method

`__init__()` is automatically called to initialize a new object. It is commonly used to set the initial values of instance attributes. It is often called a constructor in beginner courses; technically, `__new__()` creates the instance and `__init__()` initializes it.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Aman", 85)
s2 = Student("Neha", 92)

print(s1.name, s1.marks)
print(s2.name, s2.marks)
````

**Expected output**
```text
Aman 85
Neha 92
```

- `name` and `marks` are parameters.
- `self.name` and `self.marks` are attributes of the current object.
- Each instance receives its own values.

### Example: Product

````python
class Product:
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

keyboard = Product("Keyboard", 1200)
mouse = Product("Mouse", 600)

print(keyboard.product_name, keyboard.price)
print(mouse.product_name, mouse.price)
````

## 4. The `self` parameter

`self` refers to the current instance. It lets an instance method access or modify that object's attributes. `self` is a convention rather than a reserved keyword, but you should normally use this name.

When you write `fan.increase_speed()`, Python passes `fan` as the first argument automatically.

````python
class Fan:
    def __init__(self, speed):
        self.speed = speed

    def increase_speed(self):
        self.speed += 1

fan1 = Fan(2)
fan1.increase_speed()
print(fan1.speed)
````

**Expected output:** `3`

The method changes the speed of `fan1`, not every fan object.

## 5. Class attributes and instance attributes

- **Class attribute:** defined in the class body and shared as a class-level value.
- **Instance attribute:** usually assigned with `self.attribute` and belongs to an individual object.

Be careful: assigning to `object.attribute` generally creates or updates an instance attribute, even if an attribute with that name also exists on the class.

````python
class Learner:
    college = "Green Valley College"  # Class attribute

    def __init__(self, name, roll_no):
        self.name = name               # Instance attributes
        self.roll_no = roll_no

a = Learner("Anil", 101)
b = Learner("Pooja", 102)

print(a.college, a.name, a.roll_no)
print(b.college, b.name, b.roll_no)
````

**Expected output**
```text
Green Valley College Anil 101
Green Valley College Pooja 102
```

## 6. The `@property` decorator

Python's `@property` lets you access a method using attribute-style syntax. It is useful for calculated values and for validating changes through a setter.

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    @property
    def area(self):
        return self.length * self.width

rectangle = Rectangle(8, 5)
print(rectangle.area)  # No parentheses
````

**Expected output:** `40`

### Example: A property with validation

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero")
        self._celsius = value

temperature = Temperature(25)
temperature.celsius = 30
print(temperature.celsius)
````

**Expected output:** `30`

The setter checks the value before storing it. A single leading underscore, as in `_celsius`, is a convention indicating internal use; it does not make the attribute truly private.

## 7. Methods in Python classes

A **method** is a function defined inside a class.

| Type | First argument | Typical use |
|---|---|---|
| Instance method | `self` | Works with one object's data |
| Class method | `cls` | Works with class-level data |
| Static method | No automatic `self` or `cls` | Utility related to the class |

### Instance method

````python
class Calculator:
    def __init__(self, number):
        self.number = number

    def square(self):
        return self.number ** 2

calculator = Calculator(6)
print(calculator.square())
````

**Expected output:** `36`

### Class method

````python
class School:
    school_name = "Green Valley"

    @classmethod
    def change_name(cls, new_name):
        cls.school_name = new_name

School.change_name("Sunrise School")
print(School.school_name)
````

**Expected output:** `Sunrise School`

### Static method

````python
class MathTools:
    @staticmethod
    def cube(number):
        return number ** 3

print(MathTools.cube(4))
````

**Expected output:** `64`

### A method that changes object state

````python
class Wallet:
    def __init__(self, amount):
        self.amount = amount

    def add_money(self, money):
        if money > 0:
            self.amount += money

wallet = Wallet(500)
wallet.add_money(200)
print(wallet.amount)
````

**Expected output:** `700`

## 8. Inheritance

**Inheritance** allows a child class to reuse and extend functionality from a parent class.

- **Parent/base class:** the class being inherited from.
- **Child/derived class:** the class that inherits from the parent.

````python
class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    def drive(self):
        print("Car is moving")

car = Car()
car.start()  # Inherited method
car.drive()  # Child class method
````

**Expected output**
```text
Vehicle started
Car is moving
```

### Using `super()` with `__init__()`

````python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("Name:", self.name)

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def teach(self):
        print("Teaches:", self.subject)

teacher = Teacher("Meera", "Python")
teacher.introduce()
teacher.teach()
````

**Expected output**
```text
Name: Meera
Teaches: Python
```

`super().__init__(name)` calls the parent initializer.

### Multiple inheritance

````python
class Camera:
    def take_photo(self):
        print("Photo captured")

class Phone:
    def make_call(self):
        print("Calling...")

class SmartPhone(Camera, Phone):
    pass

smartphone = SmartPhone()
smartphone.take_photo()
smartphone.make_call()
````

**Expected output**
```text
Photo captured
Calling...
```

Common inheritance patterns:
1. **Single:** one child inherits from one parent.
2. **Multiple:** one child inherits from multiple parents.
3. **Multilevel:** inheritance continues through several levels.
4. **Hierarchical:** several children inherit from one parent.
5. **Hybrid:** a combination of inheritance patterns.

## 9. Polymorphism

**Polymorphism** means “many forms.” Different objects can respond to the same method name in different ways.

````python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Moo")

animals = [Dog(), Cat(), Cow()]
for animal in animals:
    animal.sound()
````

**Expected output**
```text
Bark
Meow
Moo
```

### Method overriding

````python
class Employee:
    def calculate_bonus(self):
        return 1000

class Manager(Employee):
    def calculate_bonus(self):
        return 5000

class Intern(Employee):
    def calculate_bonus(self):
        return 500

staff = [Manager(), Intern()]
for person in staff:
    print(person.calculate_bonus())
````

**Expected output**
```text
5000
500
```

Method overriding occurs when a child class provides its own implementation of a method inherited from the parent.

### Duck typing

````python
class PDF:
    def open(self):
        print("Opening PDF document")

class ImageFile:
    def open(self):
        print("Opening image")

def open_file(file):
    file.open()

open_file(PDF())
open_file(ImageFile())
````

**Expected output**
```text
Opening PDF document
Opening image
```

The function works with any object that provides a compatible `open()` method. Python often emphasizes the behavior an object supports rather than its exact class.

## 10. Encapsulation

**Encapsulation** combines data and related methods in a class and provides controlled ways to work with that data.

Python naming conventions:
- `name`: public.
- `_name`: intended for internal use (convention only).
- `__name`: name-mangled to reduce accidental clashes or access.

Double underscores do not make an attribute absolutely inaccessible; Python applies name mangling.

````python
class StudentRecord:
    def __init__(self, name):
        self.name = name
        self.__marks = 0

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")

    def get_marks(self):
        return self.__marks

student = StudentRecord("Arjun")
student.set_marks(85)
print(student.name)
print(student.get_marks())
````

**Expected output**
```text
Arjun
85
```

`__marks` is name-mangled. Access through `get_marks()` and updates through `set_marks()` provide a controlled interface.

### Example: Bank account

````python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Opening balance cannot be negative")
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit must be positive")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def show_balance(self):
        print("Balance:", self.__balance)

account = BankAccount(5000)
account.deposit(1000)
account.withdraw(2000)
account.show_balance()
````

**Expected output:** `Balance: 4000`

Encapsulation helps prevent accidental changes and allows validation to be performed through methods.

## 11. Abstraction

**Abstraction** exposes essential operations while hiding implementation details. In Python, abstract base classes can be created with the `abc` module.

````python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} by card")

class UpiPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} by UPI")

card = CardPayment()
upi = UpiPayment()

card.pay(500)
upi.pay(800)
````

**Expected output**
```text
Paid Rs. 500 by card
Paid Rs. 800 by UPI
```

- `ABC` is used to define an abstract base class.
- `@abstractmethod` declares a method that concrete subclasses must implement.
- The abstract class describes *what* an implementation should provide, while child classes define *how* it works.

### Example: Abstract shape

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

square = Square(4)
circle = Circle(3)

print(square.area())
print(round(circle.area(), 2))
````

**Expected output**
```text
16
28.27
```

### Encapsulation vs. abstraction

| Encapsulation | Abstraction |
|---|---|
| Focuses on organizing data and controlling access to it. | Focuses on hiding implementation details and exposing essential operations. |
| Example: a bank account controls balance updates. | Example: a payment interface defines `pay()` without prescribing every payment's internal process. |
| Often uses methods, properties, and naming conventions. | Often uses abstract base classes and abstract methods. |

## 12. Other useful class features

### `pass`
Use `pass` as a placeholder for a class or method that has no implementation yet.

````python
class FutureProject:
    pass

project = FutureProject()
print(type(project).__name__)
````

**Expected output:** `FutureProject`

### Updating and deleting attributes

````python
class Book:
    def __init__(self, title):
        self.title = title

book = Book("Python Basics")
book.title = "Advanced Python"
print(book.title)

del book.title
# print(book.title)  # Uncomment to see AttributeError
````

**Expected output before deletion:** `Advanced Python`

### The `__str__()` method

````python
class Movie:
    def __init__(self, title, year):
        self.title = title
        self.year = year

    def __str__(self):
        return f"{self.title} ({self.year})"

movie = Movie("The Example", 2025)
print(movie)
````

**Expected output:** `The Example (2025)`
