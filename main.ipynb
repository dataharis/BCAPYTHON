```python
#creating the class
class Student:
    name = "Vinay"
    course = "BCA"

# Creating an object
s1 = Student()

print(s1.name)
print(s1.course)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Sachin", 85)
s2 = Student("Neha", 92)

print(s1.name, s1.marks)
print(s2.name, s2.marks)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name, self.salary)

e1 = Employee("Ravi", 30000)
e2 = Employee("Sana", 45000)

e1.display()
e2.display()
```

```python
class Calculator:
    def __init__(self, number):
        self.number = number

    def square(self):
        return self.number ** 2

c = Calculator(6)
print(c.square())
```

```python
class School:
    school_name = "S K A J I C"

    @classmethod
    def change_name(cls, new_name):
        cls.school_name = new_name

School.change_name("P S S V M I C")
print(School.school_name)
```

```python
class MathTools:
    @staticmethod
    def cube(number):
        return number ** 3

print(MathTools.cube(4))
```

```python
class Wallet:
    def __init__(self, amount):
        self.amount = amount

    def add_money(self, money):
        self.amount += money

wallet = Wallet(500)
wallet.add_money(200)

print(wallet.amount)
```

```python
class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    def drive(self):
        print("Car is moving")

c = Car()
c.start()
c.drive()
```

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Moo")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def show_balance(self):
        print("Balance:", self.__balance)

account = BankAccount(5000)
account.deposit(1000)
account.withdraw(2000)
account.show_balance()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} by card")

class UpiPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} by UPI")

p1 = CardPayment()
p2 = UpiPayment()

p1.pay(500)
p2.pay(800)
```
